```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
print(A)
print(B)
```

```python
# 1. Matrix Addition
print("Addition:")
print(A + B)
```

```python
# 2. Element-wise Multiplication
print("\nElement-wise Multiplication:")
print(A * B)
```

```python
# 3. Matrix Multiplication
print("\nMatrix Multiplication:")
print(A @ B)
```

```python
# 4. Transpose
print("\nTranspose of A:")
print(A.T)
```

```python
# 5. Determinant
det_A = np.linalg.det(A)

print("\nDeterminant of A:")
print(det_A)
```

```python
## Inverse of matrix A
if np.linalg.det(A) != 0:
    A_inv = np.linalg.inv(A)
    print("Inverse:", A_inv)
    print("Verified:", np.allclose(A @ A_inv, np.eye(2)))
```

# Task 5

```python
# 1. Male passengers older than 50
result1 = df[(df["Sex"] == "male") & (df["Age"] > 50)]
print("Male passengers older than 50:", len(result1))
```

```python
# 2. Female first-class passengers
result2 = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

print("Female first-class passengers:", len(result2))
```

```python
survival_percentage = result2["Survived"].mean() * 100
print("Percentage survived:", survival_percentage)
```

```python
# 3. Age 20-40, Fare above overall median, and survived
median_fare = df["Fare"].median()
median_fare
```

```python
result3 = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > median_fare) &
    (df["Survived"] == 1)
]
print("Passengers aged 20-40 with fare above median and survived:", len(result3))
```

```python
# 4. Travelling alone, age < 30, and did not survive
result4 = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("Travelling alone, age < 30, did not survive:", len(result4))
```

```python
# 5. Embarked S, Pclass 2 or 3, fare above Southampton median
southampton = df[df["Embarked"] == "S"]
southampton_median = southampton["Fare"].median()

result5 = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]
print("Embarked S, Pclass 2 or 3, fare above Southampton median:", len(result5))
```

```python
 # 1. Survival rate by Sex
print("Survival rate by Sex:")
print(df.groupby("Sex")["Survived"].mean() * 100)
```

```python
# 2. Survival rate by Pclass
print("\nSurvival rate by Pclass:")
print(df.groupby("Pclass")["Survived"].mean() * 100)
```

```python
# 3. Average Age and Fare by Pclass
print("\nAverage Age and Fare by Pclass:")
print(df.groupby("Pclass")[["Age", "Fare"]].mean())
```

```python
# 4. Passenger count and survival rate by Sex-Pclass
print(df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
))
```

```python
# Convert survival rate to percentage
result = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)
result["Survival_Rate"] = result["Survival_Rate"] * 100
print(result)
```

```python
# 5. Passenger count, average Fare, and survival rate by Embarked
result = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)
result["Survival_Rate"] = result["Survival_Rate"] * 100
print(result)
```

# Task 7

```python
# 1. Missing values: count and percentage
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100
print(missing_count)
print(missing_percentage)
```

```python
# Plot missing counts
missing_count.plot(kind="bar")
plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Count")
plt.show()
```

```python
# Insight
# age and cabin has higher number of missing values,
# which may affect the analysis. We may need to handle these 
# missing values through imputation or removal.
```

```python
# 2. Fill missing Age with mean
print("\nAge missing before:", df["Age"].isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())

print("Age missing after:", df["Age"].isnull().sum())
```

```python
# fill missing Embarked with mean
mean_age = df["Age"].fillna(df["Age"].mean())
print(mean_age)
```

```python
# Median
median_age = df["Age"].fillna(df["Age"].median())
print(median_age)
```

```python
# Mode
mode_age = df["Age"].fillna(df["Age"].mode()[0])
print(mode_age)
```

```python
# Random value
random_age = df["Age"].fillna(
    np.random.choice(df["Age"].dropna())
)
print(random_age)
```

```python
# 4. Fare outliers using IQR

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(lower_bound)
print(upper_bound)
```

```python
# Outliers
outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]
print("Number of Outliers:", len(outliers))
```
